```python
def string_to_float_list(data_string: str):
    return np.array([float(x) for x in data_string.split()])
```

```python
lam = 635e-6 #in mm
lam_err = 5e-6

L = 144.5 #mm
L_err = 5

mm_per_pt = 0.04393
```

```python
thick_slit_widths = np.array([1.63, 2.97, 5.62, 7.11, 9.25])
thick_slit_widths_err = np.array([0.35, 0.60, 1.25, 1.30, 1.45])

y1_vals = thick_slit_widths / (2 * np.arange(len(thick_slit_widths)) + 2)
y1_vals_err = thick_slit_widths_err / (2 * np.arange(len(thick_slit_widths)) + 2)



y1_mean = np.mean(y1_vals)
y1_err = np.sqrt(np.sum(y1_vals_err**2)) / len(y1_vals_err)

#print(y1_mean, y1_err)

slit_thickness = lam * L / y1_mean

slit_thickness_err = np.sqrt((lam * L_err)**2 + (L * lam_err)**2 + (L * lam * y1_err / y1_mean)**2) / y1_mean

print(slit_thickness / mm_per_pt, slit_thickness_err / mm_per_pt)
```

```python
thin_fringe_numbers = np.array([14, 6, 8, 6, 6])

thin_fringe_distances = np.array([8.91, 3.78, 4.56, 3.32, 3.51])


thin_fringe_distances_err = np.array([1.25, 0.50, 0.50, 0.50, 0.50])

thin_fringe_width_vals = thin_fringe_distances / thin_fringe_numbers
thin_fringe_width_vals_err = thin_fringe_distances_err / thin_fringe_numbers

#print(np.round(thin_fringe_width_vals, 2))
#print(np.round(thin_fringe_width_vals_err, 2))

thin_fringe_width = np.mean(thin_fringe_width_vals)
thin_fringe_width_err = np.sqrt(np.sum(thin_fringe_width_vals_err**2)) / len(thin_fringe_width_vals_err)


print(thin_fringe_width, thin_fringe_width_err)

slit_distance = lam * L / thin_fringe_width

slit_distance_err = np.sqrt((lam * L_err)**2 + (L * lam_err)**2 + (L * lam * thin_fringe_width_err / thin_fringe_width)**2) / thin_fringe_width

print(slit_distance, slit_distance_err)

```
