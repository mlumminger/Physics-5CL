```python
d_obj = np.array([59.5, 58.0, 59.0, 58.9])
d_img = np.array([40.0, 40.1, 40.5, 40.2])

d_obj_err = 0.5
d_img_err = 0.5


f = 1 / (1/np.mean(d_obj) + 1/np.mean(d_img))

f_err = np.sqrt((d_obj_err * (f/np.mean(d_obj))**2 )**2 + (d_img_err * (f / np.mean(d_img))**2 )**2)

print(f, f_err)
```
