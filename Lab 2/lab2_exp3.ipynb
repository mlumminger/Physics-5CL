```python
theta = np.array([36, 45, 49, 55, 58, 61, 67, 70])
theta_err = 5 * np.ones(len(theta))

intensities = np.array([162.3, 728, 829, 1016, 1124, 1088, 1088, 1038])
intensities_err = np.array([0.4, 3, 1, 4, 2, 3, 3, 5])


```

```python
fig, ax = plt.subplots(figsize=(6, 4), layout="constrained")


ax.errorbar(theta, intensities, intensities_err, theta_err, fmt = 'o', markersize = 3, capsize = 4, color = 'xkcd:blue blue', zorder=10, label = "Data")



ax.set_xlabel(r"")
ax.set_ylabel(r"")


ax.grid()

ax.set_title(r"")


fig.savefig("plots/exp3_data.png", dpi=400)
```
